```python
df["Review Rating"] = df.groupby("Category")['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

## check the columns

```python
df.columns
```

```python
# here we can see the column names are not appearing properly means there is space between words and lower case upper case mixing so we need to fix it
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ',"_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns # now see the columns are appearing properly
```

```python
# create new column
labels = ['young adult','adult','middle-aged','senior']

df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df['purchase_frequency_days'] = df['purchase_frequency_days'].fillna(0).astype(int)
```

```python
df['purchase_frequency_days'] = df['purchase_frequency_days'].astype(int)
```
